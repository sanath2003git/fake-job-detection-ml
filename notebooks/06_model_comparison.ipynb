```python
metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1-Score",
    "ROC-AUC"
]

x = range(len(metrics))
width = 0.25

plt.figure(figsize=(12, 6))

plt.bar(
    [i - width for i in x],
    results_df.loc[0, metrics],
    width=width,
    label="Logistic Regression"
)

plt.bar(
    x,
    results_df.loc[1, metrics],
    width=width,
    label="Random Forest"
)

plt.bar(
    [i + width for i in x],
    results_df.loc[2, metrics],
    width=width,
    label="SVM"
)

plt.title("Overall Model Performance Comparison")
plt.xlabel("Evaluation Metric")
plt.ylabel("Score (%)")
plt.xticks(x, metrics)
plt.ylim(60, 100)
plt.legend()

plt.show()
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import numpy as np

confusion_matrices = {
    "Logistic Regression": np.array([
        [3334, 69],
        [13, 160]
    ]),
    "Random Forest": np.array([
        [3390, 13],
        [26, 147]
    ]),
    "SVM": np.array([
        [3393, 10],
        [18, 155]
    ])
}
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, (model, cm) in zip(axes, confusion_matrices.items()):
    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["Legitimate", "Fraudulent"]
    )
    disp.plot(ax=ax, cmap="Blues", colorbar=False)
    ax.set_title(model)

plt.suptitle("Confusion Matrix Comparison")
plt.tight_layout()
plt.show()
```

## Model Comparison and Findings

The three machine learning models were evaluated using Accuracy, Precision, Recall, F1-score, and ROC-AUC on the test dataset.

### Logistic Regression

Logistic Regression achieved an accuracy of 97.71%, precision of 69.87%, recall of 92.49%, F1-score of 79.60%, and ROC-AUC of 98.87%.

The model identified 160 fraudulent job postings correctly and missed 13 fraudulent postings. It produced 69 false positives.

### Random Forest

Random Forest achieved an accuracy of 98.91%, precision of 91.88%, recall of 84.97%, F1-score of 88.29%, and ROC-AUC of 99.42%.

The model correctly identified 147 fraudulent job postings and missed 26 fraudulent postings. It produced 13 false positives.

The final Random Forest evaluation used a classification threshold selected using validation-set F1-score.

### Support Vector Machine (SVM)

SVM achieved an accuracy of 99.22%, precision of 93.94%, recall of 89.60%, F1-score of 91.72%, and ROC-AUC of 98.98%.

The model correctly identified 155 fraudulent job postings and missed 18 fraudulent postings. It produced 10 false positives.

### Overall Observations

- All three models achieved high accuracy and ROC-AUC values.
- Logistic Regression produced the highest recall among the three models, identifying the largest proportion of fraudulent postings in the test set.
- Random Forest produced the highest ROC-AUC among the three models.
- SVM produced the highest recorded precision and F1-score among the three models.
- The confusion matrices show different trade-offs between false positives and false negatives.
- The dataset is imbalanced, so accuracy alone should not be used to interpret model performance.
- Precision, recall, F1-score, and ROC-AUC provide additional information about fraudulent-job detection performance.

### Preprocessing Note

The three models use the same dataset, target variable, feature groups, and stratified 80/20 train-test split. The SVM implementation uses a larger TF-IDF feature limit than the other two models, so the results should be interpreted with this preprocessing difference in mind.

## Conclusion

The experimental results show that all three traditional machine learning algorithms were able to detect fraudulent job postings with high performance. The models demonstrated different performance characteristics across accuracy, precision, recall, F1-score, and ROC-AUC.

Logistic Regression achieved the highest recall, while Random Forest achieved the highest ROC-AUC. SVM achieved the highest recorded accuracy, precision, and F1-score in the evaluated results.

These results demonstrate that traditional machine learning combined with NLP-based text representation can be used for fake job posting detection.
